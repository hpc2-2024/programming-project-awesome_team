```python
import os
NUM_THREADS = "8"
os.environ["OMP_NUM_THREADS"] = NUM_THREADS 
os.environ["OPENBLAS_NUM_THREADS"] = NUM_THREADS
os.environ["MKL_NUM_THREADS"] = NUM_THREADS
os.environ["VECLIB_MAXIMUM_THREADS"] = NUM_THREADS
os.environ["NUMEXPR_NUM_THREADS"] = NUM_THREADS
```

```python
import subprocess
import re
import numpy as np
import matplotlib.pyplot as plt
```

```python
def compile_run_program(compile_cmd, run_cmd):
    program_output = ""
    compile_process = subprocess.run(compile_cmd, capture_output=True, text=True)

    if compile_process.returncode == 0:        
        # Run the compiled program
        run_process = subprocess.run(run_cmd, capture_output=True, text=True)
        program_output = run_process.stdout
    else:
        print("Compilation failed.")
        print("Compiler output:")
        print(compile_process.stderr)
    return program_output
```

```python
# Define relative paths
SRC_DIR = ".."
PROGRAM = "mg.c"
OUTPUT = "mg.exe"
IMG_OUTPUT = "../images"
IMG = "mg_time.png"

DIMENSION="2"
SMOOTH_STEPS="2"
SMOOTHER="0"

FLAG_TIME="-time"

grid_start = 3
grid_end = 14 # max 14
GRIDSIZES = [str(2**x - 1) for x in range(grid_start, grid_end)]
LEVELS = [str(max(x-2, 2)) for x in range(grid_start, grid_end)]
iteration_counts = np.zeros((len(GRIDSIZES)))

pattern = re.compile(r"(\d+\.\d+) seconds")

compile_command = ["gcc", "-g2", "-lm", "-fopenmp", "-I../src_mg", 
                   os.path.join(SRC_DIR, PROGRAM), "-o", os.path.join(SRC_DIR, OUTPUT)]

times = np.zeros(len(GRIDSIZES))

for i, _ in enumerate(GRIDSIZES):
    GRIDSIZE = GRIDSIZES[i]
    LEVEL = LEVELS[i]

    run_command = [os.path.join(SRC_DIR, OUTPUT), DIMENSION, GRIDSIZE, LEVEL, SMOOTH_STEPS, SMOOTHER, FLAG_TIME]
    program_output = compile_run_program(compile_command, run_command)
    # print(program_output)
    match = re.search(pattern, program_output)

    if match:
        times[i] = float(match.group(1))
    else:
        times[i] = -1.0

plt.plot([int(i) for i in GRIDSIZES], times, marker='o')

plt.xscale("log", base=2)
plt.yscale("log")
plt.xlabel('Grid Size (# of points)')
plt.ylabel('Time (s)')
plt.title('Time till Convergence')

os.makedirs(IMG_OUTPUT, exist_ok=True)
output_file = os.path.join(IMG_OUTPUT, IMG)
plt.savefig(output_file)

plt.show()
```
